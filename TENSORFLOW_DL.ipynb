#**TENSORFLOW**

##TENSOR

### **Scalar (0D Tensor):**
A single number (e.g., 5).

```python
import tensorflow as tf

scalar=tf.constant(5)
print(scalar)
```

### **Vector (1D Tensor)** <br>
A vector is a one-dimensional tensor (like an array).
Example: [1, 2, 3]

```python
import tensorflow as tf

matrix=tf.constant([[1,2,3],[4,5,6]])
print(matrix)
```

### **3D Tensor**
A 3D tensor is like a collection of matrices stacked together.

```python
tensor_3d=tf.constant([[[1,2],[3,4]],[[5,6],[7,8]]])
print(tensor_3d)
```

### **Higher-Dimensional Tensors (4D, 5D, etc.)**
Used in deep learning for images, videos, and complex models.<br>
Example of a 4D Tensor: <br>
Used for image processing with shape (batch_size, height, width, channels).

```python
tensor_4d=tf.constant([[[[1],[2]], [[3],[4]]],[[[5], [6]], [[7], [8]]]])
print(tensor_4d)
```

## **TENSOR FUNCTIONS**

### **tf.constant()**
Creates a constant tensor

```python
import tensorflow as tf

tensor=tf.constant([[1,2],[3,4]]) #like array
print(tensor)
```

### **tf.zeros()**
Creates a tensor filled with zeros

```python
import tensorflow as tf
zero_tensor=tf.zeros([2,3])
print(zero_tensor)
```

### **tf.ones()**
Creates a tensor filled with ones

```python
import tensorflow as tf
one_tensor=tf.ones([2,3])
print(one_tensor)
```

### **tf.random.uniform()**

Creates a tensor with random values

```python
import tensorflow as tf
random_tensor=tf.random.uniform([2,2],minval=0,maxval=10,dtype=tf.int32)
print(random_tensor)
```

### **tf.shape()**

Returns the shape of a tensor

```python
import tensorflow as tf
tensor=tf.constant([[1,2],[3,4]])
print(tf.shape(tensor))
```

### **Mathematical Operations**

1.  tf.add() - Element-wise addition
2.  tf.subtract() - Element-wise subtraction
3. tf.multiply() - Element-wise multiplication
4. tf.divide() - Element-wise division
5. tf.matmul() - Matrix multiplication

```python
# Add

import tensorflow as tensor
a = tf.constant([1,2,3])
b = tf.constant([4,5,6])
c = tf.add(a,b)
print(c)
```

```python
# Substract

import tensorflow as tensor
a = tf.constant([1,2,3])
b = tf.constant([4,5,6])
c = tf.subtract(a,b)
print(c)
```

```python
# Multiply

import tensorflow as tensor
a = tf.constant([1,2,3])
b = tf.constant([4,5,6])
c = tf.multiply(a,b)
print(c)
```

```python
# Divide

import tensorflow as tensor
a = tf.constant([1,2,3])
b = tf.constant([4,5,6])
c = tf.divide(a,b)
print(c)
```

```python
#Matrix Multiplication

matrix1 = tf.constant([[1, 2], [3, 4]])
matrix2 = tf.constant([[5, 6], [7, 8]])
result = tf.matmul(matrix1, matrix2)
print(result)
```

### **Aggregation Functions (Summarizing Data)**

1. tf.reduce_sum() - Computes sum of all elements
2. tf.reduce_mean() - Computes mean (average)
3. tf.reduce_max() & tf.reduce_min() - Finds max/min

```python
#tf.reduce_sum()

import tensorflow as tf
tensor=tf.constant([[1,2],[3,4]])
sum=tf.reduce_sum(tensor)
print(sum)
```

```python
# tf.reduce_mean()

mean_result=tf.reduce_mean(tensor)
print(mean_result)
```

```python
# tf.reduce_min()

min_result=tf.reduce_min(tensor)
print(mean_result)
```

### **Indexing & Slicing Tensors**

```python
import tensorflow as tf
tensor=tf.constant([1,2,3,4,5])
print(tensor[0])
print(tensor[-1])
```

```python
import tensorflow as tf
tensor=tf.constant([10, 20, 30, 40, 50])
print(tensor[1:4])
```
