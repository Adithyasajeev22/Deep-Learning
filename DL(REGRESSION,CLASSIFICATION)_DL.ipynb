# **DEEP LEARNING**

Deep Learning is a method where computers learn from **large data** and make predictions.



*   It is called **“deep”** because it uses many **layers**
*   These layers help the model understand **complex patterns** like images, speech, etc



**Artificial Neural Network (ANN)**

ANN is the main model used in deep learning, inspired by the human brain.

**1. Neurons & Layers**


*   Neuron: Takes input → processes → gives output
*   Layers:

   

1.   Input Layer → receives data
2.   Hidden Layers → process data
3.   Output Layer → gives final answer


**2. Activation Functions**

They decide whether a neuron should pass its value or not

   

   



# REGRESSION 1

```python
import numpy as np
from keras.models import Sequential
from keras.layers import Dense
from keras.optimizers import Adam
```

```python
# Example: Predict y = 2x + 1 (simple linear relationship)
x_train=np.array([[1],[2],[3],[4],[5]],dtype=float)
y_train=np.array([[3],[5],[7],[9],[11]],dtype=float)
```

```python
model=Sequential([
    Dense(10,activation='relu',input_shape=(1,)), # Input layer (1 feature), hidden layer with 10 units
    Dense(1)  # Output layer for regression (1 output value)
])
```

```python
model.compile(optimizer=Adam(learning_rate=0.01),loss='mean_squared_error')
```

```python
model.fit(x_train,y_train,epochs=100,verbose=1)
```

```python
x_test=np.array([[6],[7]],dtype=float)
y_pred=model.predict(x_test)

print("Prediction for 6 and 7 : ",y_pred)
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error

# Your test data
x_test=np.array([[6],[7]],dtype=float)
y_test=np.array([[13],[15]],dtype=float)  # True values for 6 and 7 based on y=2x+1

# predict using your model
y_pred=model.predict(x_test)

# Calculate MSE and MAE
mse = mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)

print(f"Predictions: {y_pred.flatten()}")
print(f"Mean Squared Error: {mse}")
print(f"Mean Absolute Error: {mae}")
```

# REGRESSION 2

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import SGD

# Input and output data
x = np.array([[0], [1], [2], [3], [4], [5]], dtype=float)
y = np.array([[0], [0.5], [1], [1.5], [2], [2.5]], dtype=float)
```

```python
# Model
model=Sequential([Dense(1,activation='linear',input_shape=(1,))])
```

```python
# Compile with regression metrics
model.compile(optimizer=SGD(learning_rate=0.1),
              loss='mean_squared_error',
              metrics=['mean_absolute_error'])
```

```python
# Train
model.fit(x, y, epochs=200, verbose=1)
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error

# Your test data
x_test=np.array([[6],[7]],dtype=float)
y_test=np.array([[3],[4.5]],dtype=float)  # True values for 6 and 7 based on y=2x+1

# predict using your model
y_pred=model.predict(x_test)

# Calculate MSE and MAE
mse = mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)

print(f"Predictions: {y_pred.flatten()}")
print(f"Mean Squared Error: {mse}")
print(f"Mean Absolute Error: {mae}")
```

# **CLASSIFICATION**

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Generating synthetic data
x,y=make_classification(n_samples=1000,n_features=20,n_classes=2,random_state=42)

# Preprocessing: Scaling the data
scaler=StandardScaler()
x_scaled=scaler.fit_transform(x)

# Splitting data into training and testing
x_train,x_test,y_train,y_test=train_test_split(x_scaled,y,test_size=0.3,random_state=42)
```

```python
# Defining the ANN model
model = Sequential()
model.add(Dense(64, input_dim=20, activation='relu'))  # Hidden Layer 1
model.add(Dense(32, activation='relu'))               # Hidden Layer 2
model.add(Dense(1, activation='sigmoid'))             # Output Layer

# # Compiling the model
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
# Training the model
model.fit(x_train,y_train,epochs=10,batch_size=32)
```

```python
# Test prediction
# Need to add a batch dimension to the single sample X_test[0]
probs=model.predict(np.expand_dims(x_test[0],axis=0))

pred_classes=np.round(probs).astype(int)

print("Probabilities:\n", probs.flatten())
print()
print("Predicted classes (by rounding):", pred_classes.flatten())
```

```python
# Evaluating the model
loss, accuracy = model.evaluate(x_test, y_test)
print(f"Model Accuracy: {accuracy * 100:.2f}%")
```

**# KERAS MODELS**

## **1. SEQUENTIOL APL**

One Inupt and One Output(commonly usd)

```python
from keras.models import Sequential
from keras.layers import Dense,Activation
model=Sequential()
model.add(Dense(units=64,input_dim=100))
model.add(Activation('relu'))
model.add(Dense(units=10))
model.add(Activation('softmax'))
```

## **2. FUNCTIONAL API**

Multiple Inputs and Multiple Outputs

```python
from keras.layers import Input,Dense,concatenate
from keras.models import Model

input1=Input(shape=(100,))
input2=Input(shape=(50,))

hidden1=Dense(64,activation='relu')(input1)
hidden2=Dense(32,activation='relu')(input2)

merged=concatenate([hidden1,hidden2])

output=Dense(10,activation='softmax')(merged)
model=Model(inputs=[input1,input2],outputs=output)
```
